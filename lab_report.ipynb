```python
import numpy as np
import matplotlib.pyplot as plt

# Data
categories = ['A', 'B', 'C', 'D']
values = np.array([10, 20, 15, 25])

# Create bar chart
plt.bar(categories, values, color='brown')

# Labels and title
plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Simple Bar Chart")

# Show plot
plt.show()
```

```python
# Create a list of 10 integers
numbers = [12, 5, 8, 20, 3, 15, 7, 1, 18, 10]

print("Original list:", numbers)

# Append an element
numbers.append(25)
print("After appending 25:", numbers)

# Remove an element (remove value 8)
numbers.remove(8)
print("After removing 8:", numbers)

# Ascending order (sorting)
ascending = sorted(numbers)
print("Ascending order:", ascending)

# Descending order
descending = sorted(numbers, reverse=True)
print("Descending order:", descending)

# Minus operation (subtract 2 from each element)
minus_list = [x - 2 for x in numbers]
print("After subtracting 2 from each element:", minus_list)

# Find maximum value
max_value = max(numbers)
print("Maximum value:", max_value)

# Find minimum value
min_value = min(numbers)
print("Minimum value:", min_value)
```

```python
# Initialize a tuple
my_tuple = (1, 2, 3, 4, 5)

print("Original tuple:", my_tuple)

# Trying to change it will raise an error
try:
    my_tuple[0] = 10
except TypeError as e:
    print(f"Error: {e}") # Output: 'tuple' object does not support item assignment

# You can still perform non-destructive operations
print("Sum of tuple:", sum(my_tuple))
print("Length:", len(my_tuple))

# If you REALLY need to change it, you must convert it to a list first
temp_list = list(my_tuple)
temp_list.append(6)
my_tuple = tuple(temp_list)

print("New tuple after conversion:", my_tuple)
```

```python
# 1. Create a dictionary
user_data = {
    "name": "Alice",
    "age": 25,
    "city": "New York"
}
print("Initial Dictionary:", user_data)

# 2. Update a value (or add a new key-value pair)
# If the key exists, it updates; if not, it creates it.
user_data["age"] = 26          # Updating existing key
user_data["email"] = "alice@example.com"  # Adding new key
print("After Update:", user_data)

# 3. Remove a key-value pair
# Method A: Using 'del' (removes the key entirely)
if "city" in user_data:
    del user_data["city"]

# Method B: Using '.pop()' (removes the key and returns the value)
removed_val = user_data.pop("email", "Not Found")
print(f"Removed value: {removed_val}")

print("Final Dictionary:", user_data)
```
